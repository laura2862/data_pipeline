```python
import pandas as pd
from pathlib import Path

# ============================================================
# CONFIG
# ============================================================

cache_folder = Path(r"C:\Users\s7909996\Downloads\E2X\temp")

# ============================================================
# REQUIRED COLUMNS
# ============================================================

fen_cols = [
    "LegalEntityId",
    "LegalEntityName",
    "ReferenceId",
    "DocumentName",
    "iManage_Doc_Num"
]

im_cols = [
    "c1alias",
    "C_DESCRIPT",
    "docname",
    "docnum"
]

# ============================================================
# LOAD DATA
# ============================================================

fen = fen_client_doc[fen_cols].copy()
im = im_client_doc[im_cols].copy()

fen = fen.reset_index(drop=True)
im = im.reset_index(drop=True)

fen["__fen_row_id"] = fen.index
im["__im_row_id"] = im.index

# ============================================================
# CLEAN FUNCTIONS
# ============================================================

def clean_key(x):

    if pd.isna(x):
        return ""

    x = str(x).strip()

    if x.endswith(".0"):
        x = x[:-2]

    return x

def clean_text(x):

    if pd.isna(x):
        return ""

    return str(x).strip().upper()

# ============================================================
# HELPER COLUMNS
# ============================================================

fen["__key_iManage_Doc_Num"] = fen["iManage_Doc_Num"].apply(clean_key)
fen["__key_LegalEntityId"] = fen["LegalEntityId"].apply(clean_key)
fen["__key_ReferenceId"] = fen["ReferenceId"].apply(clean_key)

fen["__txt_DocumentName"] = fen["DocumentName"].apply(clean_text)
fen["__txt_LegalEntityName"] = fen["LegalEntityName"].apply(clean_text)

im["__key_docnum"] = im["docnum"].apply(clean_key)
im["__key_c1alias"] = im["c1alias"].apply(clean_key)

im["__txt_docname"] = im["docname"].apply(clean_text)
im["__txt_C_DESCRIPT"] = im["C_DESCRIPT"].apply(clean_text)

# ============================================================
# DOCUMENT MATCHING
# ============================================================

matched_fen_ids = set()
match_frames = []

# ------------------------------------------------------------
# Layer 1
# iManage_Doc_Num = docnum
# ------------------------------------------------------------

layer1 = fen.merge(
    im[
        ["__im_row_id", "__key_docnum"]
    ],
    left_on="__key_iManage_Doc_Num",
    right_on="__key_docnum",
    how="inner"
)

layer1 = layer1.drop_duplicates(
    subset=["__fen_row_id"],
    keep="first"
)

if not layer1.empty:

    tmp = layer1[
        ["__fen_row_id", "__im_row_id"]
    ].copy()

    tmp["DocMatchBoolean"] = 1
    tmp["DocMatchedBy"] = "Layer1_iManageDocNum"
    tmp["DocMatchLayer"] = 1

    match_frames.append(tmp)

    matched_fen_ids.update(
        tmp["__fen_row_id"].tolist()
    )

print("Layer 1 =", len(layer1))

# ------------------------------------------------------------
# Layer 2
# LegalEntityId = c1alias
# DocumentName = docname
# ------------------------------------------------------------

unmatched_fen = fen[
    ~fen["__fen_row_id"].isin(matched_fen_ids)
]

layer2 = unmatched_fen.merge(
    im[
        [
            "__im_row_id",
            "__key_c1alias",
            "__txt_docname"
        ]
    ],
    left_on=[
        "__key_LegalEntityId",
        "__txt_DocumentName"
    ],
    right_on=[
        "__key_c1alias",
        "__txt_docname"
    ],
    how="inner"
)

layer2 = layer2.drop_duplicates(
    subset=["__fen_row_id"],
    keep="first"
)

if not layer2.empty:

    tmp = layer2[
        ["__fen_row_id", "__im_row_id"]
    ].copy()

    tmp["DocMatchBoolean"] = 1
    tmp["DocMatchedBy"] = "Layer2_LegalEntityId_DocumentName"
    tmp["DocMatchLayer"] = 2

    match_frames.append(tmp)

    matched_fen_ids.update(
        tmp["__fen_row_id"].tolist()
    )

print("Layer 2 =", len(layer2))

# ------------------------------------------------------------
# Layer 3
# ReferenceId = c1alias
# DocumentName = docname
# ------------------------------------------------------------

unmatched_fen = fen[
    ~fen["__fen_row_id"].isin(matched_fen_ids)
]

layer3 = unmatched_fen.merge(
    im[
        [
            "__im_row_id",
            "__key_c1alias",
            "__txt_docname"
        ]
    ],
    left_on=[
        "__key_ReferenceId",
        "__txt_DocumentName"
    ],
    right_on=[
        "__key_c1alias",
        "__txt_docname"
    ],
    how="inner"
)

layer3 = layer3.drop_duplicates(
    subset=["__fen_row_id"],
    keep="first"
)

if not layer3.empty:

    tmp = layer3[
        ["__fen_row_id", "__im_row_id"]
    ].copy()

    tmp["DocMatchBoolean"] = 1
    tmp["DocMatchedBy"] = "Layer3_ReferenceId_DocumentName"
    tmp["DocMatchLayer"] = 3

    match_frames.append(tmp)

    matched_fen_ids.update(
        tmp["__fen_row_id"].tolist()
    )

print("Layer 3 =", len(layer3))

# ------------------------------------------------------------
# Layer 4
# LegalEntityName = C_DESCRIPT
# DocumentName = docname
# ------------------------------------------------------------

unmatched_fen = fen[
    ~fen["__fen_row_id"].isin(matched_fen_ids)
]

layer4 = unmatched_fen.merge(
    im[
        [
            "__im_row_id",
            "__txt_C_DESCRIPT",
            "__txt_docname"
        ]
    ],
    left_on=[
        "__txt_LegalEntityName",
        "__txt_DocumentName"
    ],
    right_on=[
        "__txt_C_DESCRIPT",
        "__txt_docname"
    ],
    how="inner"
)

layer4 = layer4.drop_duplicates(
    subset=["__fen_row_id"],
    keep="first"
)

if not layer4.empty:

    tmp = layer4[
        ["__fen_row_id", "__im_row_id"]
    ].copy()

    tmp["DocMatchBoolean"] = 1
    tmp["DocMatchedBy"] = "Layer4_LegalEntityName_DocumentName"
    tmp["DocMatchLayer"] = 4

    match_frames.append(tmp)

print("Layer 4 =", len(layer4))

# ============================================================
# BUILD MATCH TABLE
# ============================================================

if len(match_frames):

    match_df = pd.concat(
        match_frames,
        ignore_index=True
    )

    match_df = (
        match_df
        .sort_values(
            ["__fen_row_id", "DocMatchLayer"]
        )
        .drop_duplicates(
            subset=["__fen_row_id"],
            keep="first"
        )
    )

else:

    match_df = pd.DataFrame(
        columns=[
            "__fen_row_id",
            "__im_row_id",
            "DocMatchBoolean",
            "DocMatchedBy",
            "DocMatchLayer"
        ]
    )

# ============================================================
# BUILD FINAL RESULT
# ============================================================

result = fen.merge(
    match_df,
    on="__fen_row_id",
    how="left"
)

result = result.merge(
    im[
        [
            "__im_row_id",
            "c1alias",
            "C_DESCRIPT",
            "docname",
            "docnum"
        ]
    ],
    on="__im_row_id",
    how="left"
)

result["DocMatchBoolean"] = (
    result["DocMatchBoolean"]
    .fillna(0)
    .astype(int)
)

result["DocMatchedBy"] = (
    result["DocMatchedBy"]
    .fillna("No Match")
)

result["DocMatchLayer"] = (
    result["DocMatchLayer"]
    .fillna(0)
    .astype(int)
)

# ============================================================
# SAVE OUTPUT
# ============================================================

output_file = cache_folder / "fen_to_im_document_exact_match.csv"

result.to_csv(
    output_file,
    index=False
)

# ============================================================
# SUMMARY
# ============================================================

print("\nOutput:", output_file)
print("Rows:", len(result))

print("\nMatch Summary:")
display(
    result.groupby(
        ["DocMatchBoolean", "DocMatchedBy"],
        dropna=False
    )
    .size()
    .reset_index(name="Count")
    .sort_values(
        ["DocMatchBoolean", "Count"],
        ascending=[False, False]
    )
)

display(result.head())
```
